# **UNIVARIATE ANALYSIS**
Menganalisis tiap masing - masing fitur secara individu

```python
import matplotlib.pyplot as plt

for column in df.columns:
  plt.figure(figsize=(15, 6))  # Atur lebar dan panjang diagram
  plt.hist(df[column], bins=20)  # Buat histogram dengan 20 bins
  plt.title(f"Histogram of {column}")
  plt.xlabel(column)
  plt.ylabel("Frequency")
  plt.show()
```

```python
for column in df.columns:
  plt.figure(figsize=(10, 6))
  plt.scatter(df.index, df[column])
  plt.title(f"Scatter Plot of {column}")
  plt.xlabel("Index")
  plt.ylabel(column)
  plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

# Mengasumsikan dataset adalah df dan kolom 'diagnosa' adalah kategorik
dataolahencoding = df.copy() # Membuat salinan dataset untuk mencegah modifikasi dataset asli

# Label encoding untuk kolom 'diagnosa'
encoder = LabelEncoder()
dataolahencoding['diagnosa_encod'] = encoder.fit_transform(dataolahencoding['diagnosa_penyakit'])
dataolahencoding['institusi_encod'] = encoder.fit_transform(dataolahencoding['institusi'])
dataolahencoding['area_encod'] = encoder.fit_transform(dataolahencoding['area_kota'])
dataolahencoding['rumah_encod'] = encoder.fit_transform(dataolahencoding['rumah_sakit_poliklinik'])
dataolahencoding['tipe_encod'] = encoder.fit_transform(dataolahencoding['tipe_dokter'])
dataolahencoding['biaya_berobat_encod'] = encoder.fit_transform(dataolahencoding['total_biaya_penggantian'])
```

```python
dataolahencoding.head()
```

```python
dataolahencoding = dataolahencoding.drop(['diagnosa_penyakit', 'institusi', 'area_kota', 'rumah_sakit_poliklinik', 'tipe_dokter', 'total_biaya_penggantian'], axis=1)
dataolahencoding.head()
```

```python
for column in dataolahencoding.columns:
  plt.figure(figsize=(10, 6))
  plt.boxplot(dataolahencoding[column])
  plt.title(f"Box Plot of {column}")
  plt.ylabel(column)
  plt.show()
```

```python
import numpy as np

def find_outliers_iqr(data):
  Q1 = np.percentile(data, 25)
  Q3 = np.percentile(data, 75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = [x for x in data if x < lower_bound or x > upper_bound]
  return outliers

for column in dataolahencoding.columns:
  outliers = find_outliers_iqr(dataolahencoding[column])
  if outliers:
    print(f"Kolom '{column}' memiliki outlier:")
    print(outliers)
    print()
```

# **MULTIVARIATE ANALYSIS**
Menganalisis keseluruhan fitur bersama - sama

```python
import seaborn as sns

correlation_matrix = dataolahencoding.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

### **Ringkasan Insight & Penjelasan Diagram**

#### **1. Analisis Univariat (Histogram & Scatter Plot)**
* **Diagnosa Penyakit & Rumah Sakit/Poliklinik:** Terdapat dominasi data pada kategori tertentu. Data sangat menumpuk pada satu atau beberapa diagnosa (seperti hiperkolesterol/hiperlipidemia) dan faskes tertentu (seperti RS Borromeus dan Laboratorium).
* **Institusi:** Poliklinik dan Rumah Sakit adalah tempat berobat yang paling sering digunakan oleh pasien, disusul oleh Dokter Langganan.
* **Area Kota:** Sebagian besar pasien terkonsentrasi di Kota **Bandung**, dengan sebagian kecil berada di Garut, sehingga distribusi wilayahnya sangat timpang.
* **Tipe Dokter:** Pasien jauh lebih sering ditangani oleh **Dokter Umum** dibandingkan dokter spesialis lainnya.

#### **2. Analisis Outlier (Box Plot & IQR)**
* Hasil Box Plot pada data setelah di-encoding menunjukkan bahwa kolom **`area_encod`**, **`rumah_encod`**, dan **`tipe_encod`** memiliki banyak pencilan (*outliers*).
* Hal ini wajar karena sebagian besar data berpusat di satu nilai dominan (misal: area Bandung, dokter umum, RS Borromeus). Data dari kota lain atau spesialis langka otomatis terdeteksi sebagai outlier secara statistik karena frekuensinya yang sangat kecil.

#### **3. Analisis Multivariat (Correlation Matrix)**
* **Hubungan Sangat Kuat (Korelasi ~0.79):** Terdapat korelasi positif yang sangat kuat antara **`institusi_encod`** dan **`rumah_encod`**. Hal ini logis karena jenis institusi (Rumah Sakit, Poliklinik, Lab) secara langsung menentukan nama tempat faskes tersebut.
* **Korelasi Lemah/Sangat Rendah:** Fitur seperti **`biaya_berobat_encod`** hampir tidak memiliki hubungan linear dengan fitur lainnya (nilai korelasi mendekati 0 atau negatif tipis). Ini menandakan biaya pengobatan dipengaruhi oleh faktor kompleks lainnya atau memerlukan transformasi data lebih lanjut.
